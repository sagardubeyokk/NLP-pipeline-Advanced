# Final NLP Pipeline

```python
import re
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download("stopwords")
nltk.download("punkt_tab")
nltk.download("wordnet")

stop_words = set(stopwords.words("english"))
lemma = WordNetLemmatizer()

df = pd.read_csv("product_reviews.csv")
```

```python
def nlp_preprocess(text):
    # Lowercasing
    text = text.lower()

    # Noise removal 
    text = re.sub(r"http\S+|www\.\S+", "", text)           # URLs
    text = re.sub(r"[\w\.-]+@[\w\.-]+\.\w+", "", text)     # emails
    text = re.sub(r"<[^>]+>", "", text)                    # HTML tags
    text = re.sub(r"[^a-z\s]", "", text)                   # punctuation, emojis, numbers
    text = re.sub(r"\s+", " ", text).strip()               # extra spaces

    # 3. Stopword removal
    words = [w for w in text.split() if w not in stop_words]
    text = " ".join(words)

    # 4. Tokenization
    tokens = word_tokenize(text)

    # 5. Lemmatization
    tokens = [lemma.lemmatize(t) for t in tokens]

    return " ".join(tokens)
```

```python
sample = df["text"][0]
print("Raw  :", sample)
print("Final:", nlp_preprocess(sample))
```

```python
df["final_clean_text"] = df["text"].apply(nlp_preprocess)

# comparision

pd.set_option("display.max_colwidth", None)
print(df[["text", "final_clean_text"]].head(10))
```

```python
# Saving the final output

df[["text", "final_clean_text"]].to_csv("final_output.csv", index=False) 
```
