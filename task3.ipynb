# Advance Text Cleaning

```python
import pandas as pd 
import re

df = pd.read_csv("product_reviews.csv")
df["clean_text_advanced"] = df["text"]   
```

```python
# Removing URL 

df["clean_text_advanced"] = df["clean_text_advanced"].apply(
    lambda x: re.sub(r"http\S+|www\.\S+", "", x))
```

```python
df.head()
```

```python
# Removing email addresses

df["clean_text_advanced"] = df["clean_text_advanced"].apply(
    lambda x: re.sub(r"[\w\.-]+@[\w\.-]+\.\w+", "", x)
)
```

```python
df.head(10)
```

```python
# removing HTML tags 

df['clean_text_advanced']=df['clean_text_advanced'].apply(lambda x:  re.sub( r"<[^>]+>", "", x))
```

```python
print(df["clean_text_advanced"].iloc[[2, 6, 12, 19, 24, 33]].apply(repr))
```

```python
df.head()
```

```python
# Removing Special Character or emojis

df['clean_text_advanced']=df['clean_text_advanced'].apply(lambda x:re.sub(r"[^\w\s]", "", x))
```

```python
df.head()
```

```python
df["clean_text_advanced"] = df["clean_text_advanced"].str.lower() # for lowercasing 

df["clean_text_advanced"] = df["clean_text_advanced"].apply( # for removing numbers 
    lambda x: re.sub(r"\d+", "", x))

df["clean_text_advanced"] = df["clean_text_advanced"].apply(  # removing extra space 
    lambda x: re.sub(r"\s+", " ", x).strip())
```
