# Stemming

```python
import pandas as pd 
```

```python
df=pd.read_csv('task4_output.csv')
```

```python
# porter stemmer

from nltk.stem import PorterStemmer
ps=PorterStemmer()
```

```python

def stem_text(text):
    words = text.split()
    stemmed = [ps.stem(w) for w in words]
    return " ".join(stemmed)

df["stemmed_text"] = df["text_no_stopwords"].apply(stem_text)
print(df[["text_no_stopwords", "stemmed_text"]].head(5))
```

```python
all_words = " ".join(df["text_no_stopwords"]).split()
unique_words = sorted(set(all_words))

compare = pd.DataFrame({
    "original": unique_words,
    "stemmed": [ps.stem(w) for w in unique_words]
})

changed = compare[compare["original"] != compare["stemmed"]]

print("Total unique words:", len(unique_words))
print("Words changed by stemmer:", len(changed))
print(changed.head(15))
```
