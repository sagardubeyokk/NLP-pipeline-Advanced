# Basic text Cleaning 

```python
import pandas as pd 
```

```python
df=pd.read_csv('product_reviews.csv')
```

```python
df.head(5)
```

```python
df["clean_text_basic"] = df["text"].str.lower() # change all the mixed(upper and lower) data in lowercase format 
```

```python
df.head()  # for comparision
```

```python
# removing puntuation 

import re # regular expression 

df['clean_text_basic']=df['clean_text_basic'].apply(lambda x:re.sub(r"[^\w\s]","" ,x))
```

```python
df.head() # for comparision 
```

```python
# Removing Numbers

df['clean_text_basic']=df['clean_text_basic'].apply(lambda x:re.sub(r"\d+", "", x))
```

```python
df.head()
```

```python
# removing whitespace

df["clean_text_basic"] = df["clean_text_basic"].apply(lambda x: re.sub(r"\s+", " ", x).strip())
```

```python
df.head()
```

```python
pd.set_option("display.max_colwidth", None)
print(df[["text", "clean_text_basic"]].head(5))
```

''' Task 2: Basic Cleaning Steps
1. Lowercase: makes "Good" and "good" the same word.
2. Remove punctuation: removes noise like !!!, ..., :)
3. Remove numbers: removes years, prices, phone numbers.
4. Remove extra spaces: keeps one space between words.
Result stored in the column: clean_text_basic '''
