```python
import pandas as pd
import spacy

nlp = spacy.load("en_core_web_sm")

df = pd.read_csv("task4_output.csv")
```

```python
# word tokenization

def spacy_word_tokens(text):
    doc = nlp(text)
    return [token.text for token in doc]

df["word_tokens"] = df["clean_text_advanced"].apply(spacy_word_tokens)
print(df["word_tokens"][:3])
```

```python
# sent tokenization

def spacy_sent_tokens(text):
    doc = nlp(text)
    return [sent.text for sent in doc.sents]

df["sentence_tokens"] = df["text"].apply(spacy_sent_tokens)
print(df["sentence_tokens"][3])
```

```python
# comparision

for i in range(3):
    print("Original :", df["text"][i])
    print("Words    :", df["word_tokens"][i])
    print("Sentences:", df["sentence_tokens"][i])
    print()
```

# Tokenization with NLTK 

```python
import pandas as pd

df = pd.read_csv("task4_output.csv")
```

```python
df.head()
```

```python
import nltk 
nltk.download("punkt_tab")
```

```python
# word tokenize

from nltk.tokenize import word_tokenize
df['word_token']=df['clean_text_advanced'].apply(word_tokenize)
```

```python
print(df[['word_token','clean_text_advanced']].head())
```

```python
# Sent tokenize

from nltk.tokenize import sent_tokenize

df['sent_tokenize']=df['clean_text_advanced'].apply(sent_tokenize)
```

```python
print(df[['sent_tokenize','clean_text_advanced']].head())
```
