# Handling stopwords 

```python
import pandas as pd 
df = pd.read_csv("task3_output.csv")
```

# Load stopwords with NLTK

```python
import nltk
from nltk .corpus import stopwords
nltk.download('stopwords')
eng=stopwords.words('english')
```

```python
eng = set(stopwords.words('english'))
print(len(eng))
```

```python
df.head()
```

```python
# Removing stop words 

def remove_stopwords(text):
    words = text.split()                              # break the sentence into words
    kept = [w for w in words if w not in eng]         # keep only the non-stopwords
    return " ".join(kept)                             # join the words back into a sentence

df["text_no_stopwords"] = df["clean_text_advanced"].apply(remove_stopwords)
```

```python
def remove_stopwords(text):
    L = []
    for i in text.split():
        if i not in eng:
            L.append(i)
    return " ".join(L)

df["text_no_stopwords"] = df["clean_text_advanced"].apply(remove_stopwords)
```

```python
df[["text", "clean_text_advanced", "text_no_stopwords"]].to_csv("task4_output.csv", index=False)
```


**Observation:** Stopword removal makes text shorter and keeps the main words.
But "not" is also a stopword, so "not good" becomes "good" and the meaning flips.
For sentiment tasks, negation words (not, no, never) should be kept.
