# Repeted characters

#### created a small dataset for repetation and slang tasks


```python
import pandas as pd

demo = pd.DataFrame({"text": [
    "soooo good phone",
    "i loooove this product",
    "baaad quality very baaad",
    "u should buy this its gr8",
    "thx for fast delivery gr8 service",
    "plz refund my money asap",
    "the camera is sooo nice",
]})
```

```python
demo
```

```python
import re
```

```python
def fix_repeated(text):
    return re.sub(r"(.)\1{2,}", r"\1", text)

demo["step1_repeated"] = demo["text"].apply(fix_repeated)
print(demo[["text", "step1_repeated"]])
```

```python
# slang handling 
```

```python
# handwritten dict

slang_dict = {
    "u": "you",
    "gr8": "great",
    "thx": "thanks",
    "plz": "please",
    "asap": "as soon as possible",
    "r": "are",
    "its": "it is",
}
```

```python
def replace_slang(text):
    words = text.split()
    new_words = []
    for w in words:
        if w in slang_dict:
            new_words.append(slang_dict[w])
        else:
            new_words.append(w)
    return " ".join(new_words)

demo["step2_slang"] = demo["step1_repeated"].apply(replace_slang)
print(demo[["text", "step1_repeated", "step2_slang"]])
```
