# Lemmatization

```python
import pandas as pd
```

```python
df=pd.read_csv('task4_output.csv')

# no stop word column is a optimal column for lemmatization
df['text_no_stopwords'].head()
```

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.stem import WordNetLemmatizer
lemma=WordNetLemmatizer()

def lemmatize(text):
    word=text.split()
    lemmas=[lemma.lemmatize(lemma.lemmatize(w, pos="v"), pos="n") for w in word]
    return " " .join(lemmas) 

df['lemmatize_text']=df['text_no_stopwords'].apply(lemmatize)

print(df[['text_no_stopwords','lemmatize_text']].head(10))
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()
```

```python
all_words = " ".join(df["text_no_stopwords"]).split()
unique_words = sorted(set(all_words))

compare = pd.DataFrame({
    "original": unique_words,
    "stemmed": [ps.stem(w) for w in unique_words],
    "lemmatized": [lemma.lemmatize(w) for w in unique_words],
})
different = compare[compare["stemmed"] != compare["lemmatized"]]

print("Total unique words:", len(compare))
print("Words where stem and lemma differ:", len(different))
print(different.head(15))
```
