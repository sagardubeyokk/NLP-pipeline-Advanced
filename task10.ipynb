## Task 10: Observations & Insights

### 1. Difference between basic and advanced cleaning

Basic cleaning only fixed the simple things: lowercase, punctuation, numbers
and extra spaces. It worked, but it left some mess behind. For example,
`<p>Great product</p>` became `pgreat productp` because the tags were broken
into pieces instead of being removed. Links and emails also turned into
strange words like `httpsshopcom` and `johngmailcom`.

Advanced cleaning removed these things properly. URLs, emails and HTML tags
were removed first, and punctuation was removed after that. The same review
became `great product worth happy amazon`.

One thing I learned is that the order of the steps matters. If punctuation
is removed first, URLs and emails can no longer be recognised.

### 2. Why lemmatization is preferred over stemming

Stemming cuts the end of a word using fixed rules. It is fast, but it often
gives words that do not exist, like `battery -> batteri`, `quality -> qualiti`
and `happy -> happi`.

Lemmatization looks the word up in a dictionary, so the result is a real word:
`battery` stays `battery`, `days -> day`, `runs -> run`. This keeps the text
readable and keeps the meaning.

Lemmatization is slower, and WordNet treats every word as a noun by default,
so `loved` was not changed to `love` until I gave it pos="v". Even so, I
think it is the better choice when meaning matters.

### 3. Importance of preprocessing in NLP models

Raw text is full of noise like emojis, links, tags and random capital letters.
A model would treat `Good`, `GOOD` and `good` as three different words, and it
would waste effort learning from useless things like email addresses.

After preprocessing, the text is smaller, more consistent and focused on the
words that carry meaning. This gives the model cleaner input, so it can learn
better and faster.

Preprocessing also has risks. In my data, stopword removal deleted "not", so
"not good would not recommend" became "good would recommend", which flips the
meaning. So preprocessing should be matched to the task. For sentiment
analysis, negation words should be kept.

### Final pipeline
Everything is combined into one function, `nlp_preprocess(text)`. The same
steps run in the same order on every review, and the result is saved in
`final_clean_text`.
